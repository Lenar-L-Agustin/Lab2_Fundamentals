```python
target_name = "AGUSTIN"
SEED_NUM = str(7)
redacted_name = ""

for i in target_name:
    if i in "AEIOU":
        redacted_name += SEED_NUM
    else:
        redacted_name += i

print(f"Redacted name: {redacted_name}")
```

```python
ENGINEER = "AGUSTIN"
SEED_NUM = 7
first_letter = ENGINEER[0]
ASCII = ord(first_letter)
loops = SEED_NUM + 3

verification_hash = (ASCII * SEED_NUM) - loops
print(f"Forensic Signiture: {verification_hash}")
```

```python
correct_password = "agustin7"
attempts = 0

while True:
    pwd = input("Enter the password: ")
    attempts += 1
    if pwd == correct_password:
        print("Unlocked.")
        break
    else:
        print("Incorrect password. Try again.")

print(f"Input attempts: {attempts}")
```
